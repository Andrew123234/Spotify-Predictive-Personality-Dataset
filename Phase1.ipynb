```python
#imports

import pandas as pd
import numpy as np
```

```python
files = {
    "spotify_data1": "dataset.csv",
    "spotify_data2": "spotify_data.csv",
    "spotify_data3": "tracks_features.csv"
}

dataframes = {}

for name, filename in files.items():
    dataframes[name] = pd.read_csv(filename)
    print(name, len(dataframes[name]))

sp1 = len(dataframes["spotify_data1"])
sp2 = len(dataframes["spotify_data2"])
sp3 = len(dataframes["spotify_data3"])

x= sp1 + sp2 + sp3
print(x)
```

```python
spotify_data1 =  spotify_data1.rename(columns={'track_id': 'track_id', 'artist': 'artist_name'})
spotify_data2 = spotify_data2.rename(columns={'track_id': 'track_id', 'artist_name': 'artists'}) 
spotify_data3 = spotify_data3.rename(columns={'id': 'track_id', 'artist': 'artist_name', 'name': 'track_name'}) 


print(spotify_data1.columns.tolist())
print(spotify_data2.columns.tolist())
print(spotify_data3.columns.tolist())

spotify_data3["artists"] = spotify_data3["artists"].str.replace(r"[\[\]']","",regex = True)
```

```python
tables = [spotify_data1, spotify_data2, spotify_data3]
tables = [df.drop(columns=["Unnamed: 0"], errors="ignore") for df in tables]

combined = pd.concat(tables, ignore_index=True, sort=False)

combined = combined.drop_duplicates(subset=["track_id"], keep="first")
```

```python
# Keep album_name and fill its missing values from album
combined["album_name"] = combined["album_name"].combine_first(combined["album"])

# Keep genre and fill its missing values from track_genre
combined["genre"] = combined["genre"].combine_first(combined["track_genre"])

# Remove unwanted and now-redundant columns
combined = combined.drop(
    columns=["album_id", "artist_ids", "track_number", "disc_number", "album", "track_genre"],
    errors="ignore"
)
```
