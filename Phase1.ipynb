```python
#imports

import pandas as pd
import numpy as np
```

```python
spotify_data1 = pd.read_csv('dataset.csv')
spotify_data2 = pd.read_csv('spotify_data.csv')
spotify_data3 =  pd.read_csv('tracks_features.csv')

spotify_data1.head()
```

```python
spotify_data2.head()
```

```python
spotify_data1 =  spotify_data1.rename(columns={'track_id': 'track_id', 'artist': 'artist_name'})
spotify_data2 = spotify_data2.rename(columns={'track_id': 'track_id', 'artist_name': 'artist_name'}) 
spotify_data3 = spotify_data3.rename(columns={'id': 'track_id', 'artist': 'artist_name', 'name': 'track_name'}) 

spotify_data2 = spotify_data2.drop(columns=["Unnamed: 0"])

```

```python
spotify_data3["artists"] = spotify_data3["artists"].str.replace(r"[\[\]']","",regex = True)
```

```python
spotify_data3
```

```python
merged_data =  pd.merge(spotify_data1, spotify_data2, on='track_id', how='outer')
```

```python
merged_final_data = pd.merge(merged_data, spotify_data3, on='track_id', how='outer')
merged_final_data =  merged_final_data.drop_duplicates(subset=['track_id'], keep='first')
```
